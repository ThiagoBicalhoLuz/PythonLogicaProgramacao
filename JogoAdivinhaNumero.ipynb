```python
# Jogo: Adivinha o Número
# Objetivo: O usuário deve adivinhar um número aleatório entre 1 e 10

import random

# Gera o número secreto
numero_secreto = random.randint(1, 10)
tentativas = 0

print("Bem-vindo ao jogo de adivinhação!")
print("Tente adivinhar o número de 1 a 10.")

# Loop do jogo
while True:
    try:
        chute = int(input("Seu palpite: "))
        tentativas += 1

        if chute == numero_secreto:
            print(f"Parabéns! Você acertou em {tentativas} tentativas.")
            break
        elif chute < numero_secreto:
            print("Dica: O número é maior.")
        else:
            print("Dica: O número é menor.")
    except ValueError:
        print("Por favor, digite um número válido.")
```
